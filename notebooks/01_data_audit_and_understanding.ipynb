# Step 2: Create a binary column target_default

```python
# First, define mapping
default_status = [
    "Charged Off",
    "Late (31-120 days)",
    "Does not meet the credit policy. Status:Charged Off",
    "Default"
]

non_default_status = [
    "Fully Paid",
    "Does not meet the credit policy. Status:Fully Paid"
]

# Apply mapping
def map_default(status):
    if status in default_status:
        return 1
    elif status in non_default_status:
        return 0
    else:
        return np.nan  # Current, Late 16-30, In Grace, NaN

accepted_df["target_default"] = accepted_df["loan_status"].apply(map_default)

# Check counts
accepted_df["target_default"].value_counts(dropna=False)
```

```python
import pandas as pd
from google.colab import drive
import os

# Mount Drive
drive.mount('/content/drive', force_remount=True)

# Paths
PROJECT_ROOT = "/content/drive/MyDrive/LOAN_DEFAULT_RISK/"
FILTERED_PATH = PROJECT_ROOT + "data/filtered/"
PROCESSED_PATH = PROJECT_ROOT + "data/processed/"

# Ensure filtered folder exists
os.makedirs(FILTERED_PATH, exist_ok=True)

# Step 1: Load accepted + rejected CA loans
ca_df = pd.read_csv(FILTERED_PATH + "accepted_CA.csv", low_memory=False)
rejected_ca_df = pd.read_csv(FILTERED_PATH + "rejected_CA.csv", low_memory=False)

# Add loan status
ca_df['loan_status'] = 'Accepted'
rejected_ca_df['loan_status'] = 'Rejected'

# Combine datasets
full_ca_df = pd.concat([ca_df, rejected_ca_df], ignore_index=True)
print("Full CA dataset shape:", full_ca_df.shape)

# Step 2: Prepare full CA dataset for merge
full_ca_df['Application Date'] = pd.to_datetime(full_ca_df['Application Date'], errors='coerce')
full_ca_df = full_ca_df.dropna(subset=['Application Date'])
full_ca_df['year'] = full_ca_df['Application Date'].dt.year
full_ca_df['month'] = full_ca_df['Application Date'].dt.month

# Step 3: Load CA unemployment and prepare keys
unemployment_ca_df = pd.read_csv(FILTERED_PATH + "ca_unemployment_clean.csv", low_memory=False)
unemployment_ca_df['year'] = unemployment_ca_df['year_month'].str[:4].astype(int)
unemployment_ca_df['month'] = unemployment_ca_df['year_month'].str[5:7].astype(int)
unemployment_ca_df = unemployment_ca_df[['year','month','unemployment_rate']]

# Step 4: Merge in chunks to avoid memory crash
merged_chunks = []

for y, m in unemployment_ca_df[['year','month']].drop_duplicates().values:
    temp_loans = full_ca_df[(full_ca_df['year']==y) & (full_ca_df['month']==m)]
    temp_unemp = unemployment_ca_df[(unemployment_ca_df['year']==y) & (unemployment_ca_df['month']==m)]
    temp_merged = temp_loans.merge(temp_unemp, on=['year','month'], how='left')
    merged_chunks.append(temp_merged)

# Concatenate all chunks
full_ca_merged = pd.concat(merged_chunks, ignore_index=True)
print("Merged dataset shape:", full_ca_merged.shape)

# Step 5: Save final merged dataset
full_ca_merged.to_csv(FILTERED_PATH + "full_CA_merged_dataset.csv", index=False)
print("Merged dataset saved at:", FILTERED_PATH + "full_CA_merged_dataset.csv")
```
