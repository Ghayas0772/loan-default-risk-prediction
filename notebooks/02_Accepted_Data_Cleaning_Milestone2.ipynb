### 1️⃣ Mount Google Drive & Define Paths

```python
from google.colab import drive
import pandas as pd

# Mount Google Drive
drive.mount('/content/drive')

# Paths
PROJECT_ROOT = "/content/drive/MyDrive/LOAN_DEFAULT_RISK/"
RAW_PATH = PROJECT_ROOT + "data/raw/"
PROCESSED_PATH = PROJECT_ROOT + "data/processed/"

# Files
ACCEPTED_FILE = PROCESSED_PATH + "accepted_checked.csv"
COLUMN_CLASS_FILE = PROCESSED_PATH + "column_classification.csv"
```

### 2️⃣ Load Accepted Dataset & Column Dictionary

```python
# Load dataset
accepted_df = pd.read_csv(ACCEPTED_FILE)

# Load column classification
column_class_df = pd.read_csv(COLUMN_CLASS_FILE)

# Check first few rows
print("Accepted Dataset Shape:", accepted_df.shape)
accepted_df.head()
```

### 3️⃣ Separate Columns by Category

```python
# Create lists of columns
id_cols = column_class_df[column_class_df['category'] == 'ID']['column'].tolist()
leakage_cols = column_class_df[column_class_df['category'] == 'Leakage']['column'].tolist()
predictor_cols = column_class_df[column_class_df['category'] == 'Predictor']['column'].tolist()
target_col = column_class_df[column_class_df['category'] == 'Target']['column'].tolist()[0]

print("ID columns:", len(id_cols))
print("Leakage columns:", len(leakage_cols))
print("Predictor columns:", len(predictor_cols))
print("Target column:", target_col)
```

### 4️⃣ Create Modeling-Ready Dataset

```python
# Drop ID and leakage columns
accepted_model_df = accepted_df.drop(columns=id_cols + leakage_cols)

# Ensure target is included
accepted_model_df[target_col] = accepted_df[target_col]

print("Modeling-ready dataset shape:", accepted_model_df.shape)
accepted_model_df.head()
```

```python
# Ensure target column is included
if 'target_default' not in accepted_model_df.columns:
    accepted_model_df['target_default'] = accepted_df['target_default']

# Quick check
print("Columns in modeling dataset:", accepted_model_df.columns[-5:])  # last few columns
print("Shape:", accepted_model_df.shape)
```

```python
# Check target distribution
accepted_model_df['target_default'].value_counts(dropna=False)
```

```python
print(accepted_model_df)
```

### 5 Handling missing data
checking missing values

```python
missing = accepted_model_df.isna().sum().sort_values(ascending=False)
missing[missing > 0]
```

### Step 1: Handle high-missing columns

Rule of thumb:

* If a column has >50% missing, consider dropping it (especially if it’s secondary applicant info).

* Otherwise, impute numeric with median and categorical with mode.


```python
# Drop columns with >50% missing
threshold = 0.5 * len(accepted_model_df)
high_missing_cols = accepted_model_df.isna().sum()[accepted_model_df.isna().sum() > threshold].index.tolist()
print("Dropping high-missing columns:", high_missing_cols)

accepted_model_df.drop(columns=high_missing_cols, inplace=True)
```

### Step 2: Impute remaining numeric missing values
After dropping very sparse columns:

- Numeric columns → fill missing values with median

-  Categorical columns → fill missing values with mode

```python
# Remaining numeric columns (excluding target)
num_cols = accepted_model_df.select_dtypes(include=['float64','int64']).columns.tolist()
num_cols = [c for c in num_cols if c != 'target_default']

for col in num_cols:
    accepted_model_df[col] = accepted_model_df[col].fillna(accepted_model_df[col].median())

```

### Step 3: Impute remaining categorical missing values

```python
cat_cols = accepted_model_df.select_dtypes(include=['object']).columns.tolist()

for col in cat_cols:
    accepted_model_df[col] = accepted_model_df[col].fillna(accepted_model_df[col].mode()[0])
```

### Step 4: Verify all missing values are gone

```python
accepted_model_df.isna().sum().sum()
```

Fully numeric + categorical columns intact

Target column included (target_default)

IDs and leakage columns removed

No missing values → ready for encoding

### 4- Encoding categorical variables
All categorical predictors (like home_ownership, verification_status, initial_list_status, application_type, addr_state, etc.) need to be converted into numeric form for modeling.

```python
# Identify categorical columns
cat_cols = accepted_model_df.select_dtypes(include=['object']).columns.tolist()

# One-hot encode all categorical columns (drop first to avoid dummy variable trap)
accepted_model_df = pd.get_dummies(accepted_model_df, columns=cat_cols, drop_first=True)

print("After encoding shape:", accepted_model_df.shape)
```

### Save your clean modeling dataset

```python
accepted_model_df.to_csv(PROCESSED_PATH + "accepted_clean_model.csv", index=False)
print("Clean modeling dataset saved successfully!")
```
